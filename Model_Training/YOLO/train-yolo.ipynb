```python
from ultralytics import YOLO
import shutil
import os

def save_checkpoint(trainer):
    if trainer.epoch % 5 == 0:
        src = '/kaggle/working/runs/detect/train_final/weights/last.pt'
        dst = f'/kaggle/working/backup_epoch_{trainer.epoch}.pt'
        if os.path.exists(src):
            shutil.copy(src, dst)
            print(f"✅ Saved backup at epoch {trainer.epoch}")

model = YOLO('/kaggle/input/datasets/sademfeehan/yolo-checkpoint/backup_epoch_10.pt')
model.add_callback('on_train_epoch_end', save_checkpoint)

model.train(
    data='/kaggle/working/data.yaml',
    epochs=100,
    imgsz=640,
    device=0,
    batch=16,
    workers=4,
    exist_ok=True,
    name='train_final'
)
```
